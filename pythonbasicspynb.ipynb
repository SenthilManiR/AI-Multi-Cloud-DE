```python
#Python treats single quotes the same as double quotes as like triple quotes, but has some differences
company1='inceptez'
company2="inceptez"
company3="""inceptez"""
print(company1,company2,company3)#python treat all as same
```

```python
company1='''inceptez's'''
company2="inceptez's"
company3=""""inceptez's"""
print(company1,company2,company3)
```

```python
#Triple quotes (either '''/""") place very important roles
#For holding both single, double quoted data we can use triple quotes
company3="""This is "inceptez's" company's property"""
print(company3)
```
